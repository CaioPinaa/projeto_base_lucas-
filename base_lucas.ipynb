```python
# aqui eu posso usar um for in range + lista vazia + sum  para deixar o código mais enxuto , principalmente o for in range para não ter que escrever 4 linhas de provas : 

# Média das provas.
# Solicite ao usuário o seu nome, as notas de suas 4 provas, e calcule a média dessas notas.
# De acordo com a média, verifique a situação do aluno:

# Média de 0 a 4 - reprovado
# Média de 4 a 7 - recuperação
# Média de 7 a 10 - aprovado

# Ao final, mostre a seguinte mensagem:

# O(a) aluno(a) {nome} teve como média {media} e sua situação é: {situacao}

nome_usuario = input("Insira o nome do usuário aqui : ")

notas = []

for _ in range (4) : 
    nota = float(input("Insira a nota da prova aqui: "))
    notas.append(nota)

media_notas = sum(notas) / 4

if media_notas >=0 and media_notas <= 4 : 
    situacao = "Reprovado"
elif media_notas >= 4 and media_notas <= 7 :
    situacao = "Recuperação"
else : 
    situacao = "Aprovado"

print (f"O aluno {nome_usuario} teve como média a {media_notas} e sua situação é {situacao} ")

```
